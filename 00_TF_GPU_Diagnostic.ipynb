# TensorFlow GPU Diagnostic
Run this after sourcing the correct LD_LIBRARY_PATH for the venv's NVIDIA wheels.

```python
import os
# Add NVIDIA wheel libs if not already set
venv = os.getcwd() + '/.venv'
paths = [
    venv + '/lib/python3.13/site-packages/nvidia/cuda_runtime/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cudnn/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cublas/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cufft/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cusolver/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cusparse/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cuda_nvrtc/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cuda_cupti/lib',
    venv + '/lib/python3.13/site-packages/nvidia/nccl/lib',
    venv + '/lib/python3.13/site-packages/nvidia/nvjitlink/lib',
    venv + '/lib/python3.13/site-packages/nvidia/nvtx/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cufile/lib',
    venv + '/lib/python3.13/site-packages/nvidia/curand/lib',
    venv + '/lib/python3.13/site-packages/nvidia/cusparselt/lib',
    venv + '/lib/python3.13/site-packages/nvidia/nvshmem/lib',
    '/usr/local/cuda/lib64'
]
old = os.environ.get('LD_LIBRARY_PATH','')
os.environ['LD_LIBRARY_PATH'] = ':'.join(paths) + (':' + old if old else '')
print('LD_LIBRARY_PATH set')
```

```python
import tensorflow as tf
print('TensorFlow:', tf.__version__)
print('Built with CUDA:', tf.test.is_built_with_cuda())
gpus = tf.config.list_physical_devices('GPU')
print('GPUs:', gpus)
for g in gpus:
    try:
        print(tf.config.experimental.get_device_details(g))
    except Exception as e:
        print(e)
```
