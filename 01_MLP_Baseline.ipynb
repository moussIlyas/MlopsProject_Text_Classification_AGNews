# MLP Baseline (PyTorch) - AG News Classification

```python
import os
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mlflow
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from mlflow_config import configure_mlflow

configure_mlflow()
mlflow.set_experiment("AG_News_Classification")

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
```

```python
train_df = pd.read_csv('data/train.csv')
test_df = pd.read_csv('data/test.csv')
train_df['text'] = train_df['Title'] + ' ' + train_df['Description']
test_df['text'] = test_df['Title'] + ' ' + test_df['Description']
X_train, y_train = train_df['text'].values, (train_df['Class Index']-1).values
X_test, y_test = test_df['text'].values, (test_df['Class Index']-1).values
```

```python
max_words = 10000
max_len = 100

def tokenize(texts):
    tokens = []
    for t in texts:
        tokens.append(t.lower().split())
    return tokens

train_tokens = tokenize(X_train)
counter = Counter([w for s in train_tokens for w in s])
vocab = {w:i+2 for i,(w,_) in enumerate(counter.most_common(max_words-2))}
vocab['<PAD>']=0
vocab['<UNK>']=1

def encode(tokens):
    seqs = []
    for s in tokens:
        seq = [vocab.get(w,1) for w in s][:max_len]
        seq += [0]*(max_len-len(seq))
        seqs.append(seq)
    return torch.tensor(seqs, dtype=torch.long)

X_train_seq = encode(train_tokens)
X_test_seq = encode(tokenize(X_test))
y_train_t = torch.tensor(y_train, dtype=torch.long)
y_test_t = torch.tensor(y_test, dtype=torch.long)
```

```python
class NewsDataset(Dataset):
    def __init__(self,x,y):
        self.x=x; self.y=y
    def __len__(self): return len(self.y)
    def __getitem__(self,i): return self.x[i], self.y[i]

train_loader = DataLoader(NewsDataset(X_train_seq,y_train_t), batch_size=64, shuffle=True)
test_loader = DataLoader(NewsDataset(X_test_seq,y_test_t), batch_size=64)
```

```python
class MLP(nn.Module):
    def __init__(self,vocab_size,embed_dim=128,max_len=100):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.fc = nn.Sequential(
            nn.Linear(embed_dim*max_len, 256),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256,128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128,4)
        )
    def forward(self,x):
        x = self.embed(x)
        x = x.view(x.size(0), -1)
        return self.fc(x)

model = MLP(len(vocab)).to(device)
```

```python
with mlflow.start_run(run_name='MLP_Baseline_PyTorch'):
    mlflow.log_param('model_type','MLP_PyTorch')
    mlflow.log_param('max_words',max_words)
    mlflow.log_param('max_len',max_len)
    mlflow.log_param('device',str(device))
    
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    
    train_accs, val_accs = [], []
    for epoch in range(10):
        model.train()
        correct,total,loss_sum=0,0,0
        for xb,yb in train_loader:
            xb,yb=xb.to(device),yb.to(device)
            optimizer.zero_grad()
            out=model(xb)
            loss=criterion(out,yb)
            loss.backward()
            optimizer.step()
            loss_sum+=loss.item()
            pred=out.argmax(1)
            correct+=(pred==yb).sum().item()
            total+=yb.size(0)
        train_accs.append(correct/total)
        
        model.eval()
        correct,total=0,0
        with torch.no_grad():
            for xb,yb in test_loader:
                xb,yb=xb.to(device),yb.to(device)
                out=model(xb)
                pred=out.argmax(1)
                correct+=(pred==yb).sum().item()
                total+=yb.size(0)
        val_accs.append(correct/total)
        mlflow.log_metric('val_acc',val_accs[-1],step=epoch)
        print(f'Epoch {epoch+1}: train_acc={train_accs[-1]:.4f} val_acc={val_accs[-1]:.4f}')
    
    model.eval()
    y_pred=[]
    with torch.no_grad():
        for xb,yb in test_loader:
            xb=xb.to(device)
            out=model(xb)
            y_pred.extend(out.argmax(1).cpu().numpy())
    acc=accuracy_score(y_test,y_pred)
    mlflow.log_metric('test_accuracy',acc)
    
    report=classification_report(y_test,y_pred,target_names=['World','Sports','Business','Sci/Tech'])
    with open('results/MLP_Baseline_results.txt','w') as f:
        f.write(report)
        f.write(f'\nAccuracy: {acc}\n')
    mlflow.log_artifact('results/MLP_Baseline_results.txt')
    
    plt.figure(figsize=(10,4))
    plt.plot(train_accs,label='train')
    plt.plot(val_accs,label='val')
    plt.title('MLP (PyTorch) Accuracy')
    plt.legend()
    plt.savefig('Figures/MLP_Baseline_accuracy_loss.png')
    plt.close()
    mlflow.log_artifact('Figures/MLP_Baseline_accuracy_loss.png')
    
    cm=confusion_matrix(y_test,y_pred)
    plt.figure(figsize=(8,6))
    sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',xticklabels=['World','Sports','Business','Sci/Tech'],yticklabels=['World','Sports','Business','Sci/Tech'])
    plt.title('MLP Confusion Matrix')
    plt.savefig('Figures/MLP_Baseline_confusion_matrix.png')
    plt.close()
    mlflow.log_artifact('Figures/MLP_Baseline_confusion_matrix.png')
    print(report)
```
